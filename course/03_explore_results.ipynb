# 3. Explore the EcoSIM result

Read ATS observation output from the first run. The `time [s]` column is seconds since the simulation start; divide by 86,400 to plot days.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from scripts.run_case import ROOT, run_case
base_file = ROOT / "snow_dynamics_albedo_internal.demo" / "surf_prop.dat"
assert base_file.exists(), "Run notebook 02_first_run.ipynb first"
base = pd.read_csv(base_file, comment="#")
print(base.columns.tolist())
base.head()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(base["time [s]"] / 86400, base["snow_depth"] * 100, label="EcoSIM internal albedo")
ax.set(xlabel="Time (days)", ylabel="Snow depth (cm)")
ax.legend()
plt.show()
```

## Compare prescribed albedo

The next cell runs the low and high albedo inputs for the same 30-day period. Their `.demo` directories are independent, so each keeps its own XML, output, and log. This may take longer than the first run.


```python
low_dir = run_case("snow_dynamics_albedo_low", days=30)
high_dir = run_case("snow_dynamics_albedo_high", days=30)
low = pd.read_csv(low_dir / "surf_prop.dat", comment="#")
high = pd.read_csv(high_dir / "surf_prop.dat", comment="#")
fig, ax = plt.subplots(figsize=(8, 4))
for frame, label in ((base, "internal"), (low, "low"), (high, "high")):
    ax.plot(frame["time [s]"] / 86400, frame["snow_depth"] * 100, label=label)
ax.set(xlabel="Time (days)", ylabel="Snow depth (cm)")
ax.legend(title="Snow albedo")
plt.show()
```
