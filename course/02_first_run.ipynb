# 2. Your first EcoSIM run

Run a 30-day snow simulation using EcoSIM’s internal albedo. The original demo spans 1,826 days; this shorter run makes the first exercise manageable. The helper copies the input into a `.demo` directory, sets its end time to 30 days, and keeps the forcing paths valid.


```python
from scripts.run_case import prepare_case
case = "snow_dynamics_albedo_internal"
run_dir = prepare_case(case, days=30)
print(run_dir)
print("Input:", run_dir / f"{case}.xml")
```

Inspect the copied XML before running. Find the **end time**, **EcoSIM for ATS**, and **prescribe snow albedo** parameters. The original input stays in `inputs/`.


```python
from xml.etree import ElementTree as ET
tree = ET.parse(run_dir / f"{case}.xml")
for key in ("end time", "prescribe snow albedo"):
    matches = [p.get("value") for p in tree.iter("Parameter") if p.get("name") == key]
    print(key, matches[:1])
```

This step can take several minutes. Output and the full log are stored in the run directory. If ATS fails, the helper shows the last 25 log lines.
